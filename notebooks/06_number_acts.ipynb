```python
from helixadd import model as hmodel, helix
model = hmodel.load_model()
```

```python
data = helix.collect_numbers(model)
print({k: v.shape for k, v in data.items()})
```

```python
# quick look: the strongest period at each point, for a
import numpy as np
for i in range(17):
    s = helix.spectrum(data["a"][i])
    top = np.argsort(s[1:])[::-1][:3] + 1           # skip k = 0
    print("embed" if i == 0 else f"L{i - 1:>2}", "  ".join(f"T={100 / k:5.1f} ({s[k]:.2f})" for k in top))
```
